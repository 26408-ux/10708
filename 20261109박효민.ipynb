```python
from gtts import gTTS
text = "크아악 롤랑이새끼가 크윽 롤랑이 미친 시발놈이 중년 주제에 딸딸이를 부르는구나 오냐 덤벼랏 딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸딸"
tts = gTTS( text=text, lang="ko")
tts.save("output.mp3")
print ("음성파일 저장 완")
```
